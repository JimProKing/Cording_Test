# 회의실 예약
- 그리디 알고리즘 사용

1. 회의 종료시간 기준으로 정렬 (종료시간 빠른건 무조건 고르는게 좋음)
2. 안겹치는거 선택 선택 선택
3. 다 돌면 해당 선택 횟수 리턴

```python
N = int(input())
## A = [[0]*2 for _ in range(N)]
A = []
for i in range(N):
    S,E = map(int,input().split())
    A.append([E,S])
A.sort()
cnt = 0; end = -1

for i in range(N):
    if A[i][1]>=end:
        end = A[i][0]; cnt += 1
print(cnt)

# 종료시간 젤 빠른거 일단 시작
# 그 뒤 종료시간 빠른거 시작
# 계속 반복. 끝났을때 결과 출력 
```
