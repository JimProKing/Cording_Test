- 시간복잡도 때문에 이진탐색 사용하기

인풋
- N: N*N 배열
- K: K번째 수

```python
N = int(input())
K = int(input())

s = 1
e = K
ans = 0

while s <= e:
    m = (s + e) // 2
    cnt = 0

    for i in range(1, N + 1):
        cnt += min(m // i, N)

    if cnt < K:
        s = m + 1
    else:
        ans = m
        e = m - 1

print(ans)
```
