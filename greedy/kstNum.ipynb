- N: N*N 배열
- K: K번째 수

```python
N = int(input());K=int(input())
s=1;e=K;ans=0##K보다는 무조건 작음

while s <= e:
    m = int((s+e)/2);cnt=0 #중앙값보다 작은 수 계산
for i in range(1,N+1):
    cnt+=min(int(m/i),N)
if cnt<K:
    s = m+1
else:
    ans = m;e=m-1
print(ans)    
```
