```python
# 223 p
from queue import PriorityQueue
N = int(input())
pq = PriorityQueue() # 우선순위 큐 그냥 호출하면 최소힙

for _ in range(N):
    data = int(input())
    pq.put(data)

data1 = 0;data2=0;sum=0

while pq.qsize()>1:
    data1 = pq.get()
    data2 = pq.get()
    temp = data1+data2
    sum+=temp
    pq.put(temp)

print(sum)
```
