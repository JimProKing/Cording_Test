# 과정
1. 데이터 4개 그룹에 나눠서 넣음
- 양수, 음수, 1, 0
2. 양수 큐 큰것부터 2개씩 뽑아서 결과 더함 * 하나 남으면 그것도 걍 더함
3. 음수 큐 절댓값 큰것부터 2개씩 뽑아서 결과 더함. * 하나 남았을 때 0 있으면 버림. 0 남은거 없으면 더함.
4. 1 개수 더함.
5. 결과 출력

```python
from queue import PriorityQueue
N = int(input())
plusPq = PriorityQueue();minusPq =  PriorityQueue(); one = 0;zero = 0

for i in range(N):
    data = int(input())
    if data > 1:
        plusPq.put(data*-1) ## 내림차 정렬 위해 -1 곱해서
    elif data == 1:
        one+=1
    elif data == 0:
        zero+=1
    else:
        minusPq.put(data)
sum = 0
while plusPq.qsize()>1:
    first = plusPq.get();second = plusPq.get()
    sum += first*second
if plusPq.qsize()>0: ## 1개 남았다면
    sum += plusPq.get()*-1

while minusPq.qsize() > 1:
    first = minusPq.get();second = minusPq.get()
    sum  += first*second
if minusPq.qsize() > 0:
    if zero == 0:
        sum += minusPq.get()
sum += one
print(sum)
```
