# 트리의 지름
트리를 구성하는 노드 중, 두  노 사의 거리가 가장 긴 것
- -1: 더 이상 노드 없고 종료

```python
from collections import deque 

N = int(input())
A = [[] for _ in range(N+1)]

for i in range(N):
    Data = list(map(int, input().split()))
    index = 0
    S = Data[index] ## 시작점
    index +=1 ## E,V 찾기. Data[index]는 E, Data[index+1]는 V
    while True:
        E = Data[index]
        if E == -1:
            break
        V = Data[index + 1]
        A[S].append((E,V))
        index += 2
## A 구조 [[],[](2.3),(3,5)...]
distance = [0]*(N+1) ## 거리를 기록하기 위한 배열
visited = [0]*(N+1) ## 방문 여부 기록위한 배열

def BFS(v): ## LV1 다 돌고 LV2
    queue = deque()
    queue.append(v)
    visited[v] = 1
    while queue:
        now = queue.popleft()
        for i in A[now]:
            if not visited[i[0]]:
                ## 이 부분이 로직
                visited[i[0]] = 1 
                queue.append(i[0])
                distance[i[0]] = distance[now] + i[1]
BFS(1)
Max = 1
for i in range(2,N+1):
    if distance[Max] < distance[i]:
        Max = i
distance = [0]*(N+1)
visited = [0]*(N+1)
BFS(Max)
distance.sort()
print(distance[N])
```
