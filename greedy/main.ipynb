```python
# 동전 개수의 최솟값
### N종류의 동전 충분히 가질 때, 적절히 사용해 그 가격 합을 K로
### 동전 개수 최솟값?

### 동전 수 N     목표 금액 K
### 종류1
### 종류2.....

N, K = map(int,input().split())
A = [0]*N ## 동전 종류 담을거임

for i in range(N):
    A[i] = int(input())

cnt = 0

for i in range(N-1,-1,-1):
    ## 큰 것 부터
    if A[i]<=K:
        cnt += K//A[i]
        K %= A[i]
print(cnt)
```
