# 제곱이아닌 수
- 정의: 1보다 큰 제곱수로 나누어떨어지지 않는 수
- min보다 크고 max보다 작은 수 중 제곱이 아닌 수 출력

```python
# 내가 짠 코드
mn, mx = map(int, input().split())

temp = []
for i in range(mn, mx + 1):
    j = 2
    while j * j <= i:
        if i % (j * j) == 0:
            temp.append(i)
            break
        j += 1
answer = []
for i in range(mn,mx+1):
    if i not in temp:
        answer.append(i)
print(answer)
print(len(answer))
```

```python
# 정석
import math
mn, mx = map(int,input().split())
check = [0]*(mx-mn+1)

for i in range(2,int(math.sqrt(mx))+1):
    pow = i*i;start = mn//pow
    if mn%pow != 0:
        start +=1
    for j in range(start,mx//pow+1):
        check[int((j*pow)-mn)]=1
count = 0
for i in range(0,mx-mn+1):
    if not check[i]:
        count+=1
print(count)
```
