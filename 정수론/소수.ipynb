### 에라토스테네스의 채를 이용해 소수를 판별
- 이중 for문처럼 보여서 O(N^2)처럼 보임
- 실제로는 O(Nlog(logN))
  - 배수를 삭제하는 연산으로, 실제 구현에서 바깥 for문 생략 경우가 많기 때문.

```python
import math
M,N = map(int,input().split())
A = [0]*(N+1) ## 첫번째칸 안쓸거임

for i in range(2,N+1):
    A[i]=i ## 1은 소수가 아니니까

for i in range(2,int(math.sqrt(N))+1):## 2~루트N까지, 해당 수의 배수들을 다 지우는 작업 수행
    if A[i]==0:continue
    for j in range(2*i,N+1,i): ##i씩 키워가면서 배수 모두 제거
        A[j]=0
for i in range(M,N+1):
    if A[i] != 0:
        print(A[i])
        
```
