```python
# 팰린드롬 수: 숫자 뒤집은게 일치하는거
# N 이상이고 소수이면서 팰린드롬 수 인 것중에 가장 작은거

import math
N = int(input())
A = [0]*(10000001)

## A배열을 소수 배열로 바꾸기
for i in range(2,len(A)):
    A[i] = i
for i in range(2,int(math.sqrt(len(A))+1)):##제곱근까지만 수행
    if A[i] == 0:
        continue
    for j in range(i+i,len(A),i):
        A[j]=0
def isPalindrom(target):
    temp = list(str(target))
    s = 0
    e = len(temp) - 1
    while(s<e):
        if temp[s] != temp[e]:
            return False
        s+=1;e-=1
    return True

i = N
while True:
    if A[i]!=0:
        result = A[i]
        if (isPalindrom(result)):
            print(result);break
    i+=1
```
