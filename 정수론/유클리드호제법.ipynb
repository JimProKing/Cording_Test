```python
## 유클리드 호제법 - 두 수의 최대공약수 
## mod 연산 :: 두 값을 나눈 나머지 구하는 연산
## ex) 10 mod 4 = 2

# 과정
## 1. 큰 수 mod 작은 수
## 2. 작은 수 mod 1의 결과
## 3. 2반복 -- 값이 0되면 작은 수가 최대공약수

def gcd(a,b):
    if b==0:
        return a
    else:
        return gcd(b, a%b)
    
t = int(input())

for i in range(t):
    a, b = map(int,input().split())
    result = int(a*b / gcd(a,b))
    print(result)
```
