# 오일러 피 함수 :: P[N]
- 1~N 범위 중 N과 서로소인 자연수 개수

```python
N = int(input())
temp = []
for i in range(2, N+1):
    if N % i == 0:
        temp.append(i)   # N의 약수들

check = [0] * (N + 1)

for i in temp:
    j = i
    while j <= N:
        check[j] = 1
        j += i

answer = []
for i in range(1, N+1):
    if check[i] == 0:
        answer.append(i)

print(answer)
print(len(answer))

# 비효율적. 보통 소인수분해로 구함.
```
