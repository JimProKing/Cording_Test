````python
N,E = map(int,input().split())
visited = [0]*(N+1)
A = [[] for _ in range(N+1)] ## [[],[],[],[],...]
for i in range(E):
    a,b = map(int,input().split())
    ## 양방향 엣지라 양쪽에 다 
    A[a].append(b)
    A[b].append(a)

def DFS(v):
    visited[v] = 1
    for i in A[v]:
        if not visited[i]:
            DFS(i)
count = 0
for i in range(1,N+1): ## 1~n까지 수행
    if visited[i] != 1:
        count+=1
        DFS(i)
print(count)
````

### 신기한 소수 찾기
숫자 N이 주어졌을 때, N자리 소수 중 신기한 소수 모두 찾기 (1<= N <= 8)

````python
# 소수 판별 함수
def is_prime(i: int) -> bool:
    if i < 2:
        return False
    if i == 2:
        return True
    if i % 2 == 0:
        return False
    r = int(i**0.5)
    for d in range(3, r + 1, 2):
        if i % d == 0:
            return False
    return True

base = 10
N = int(input("자릿수 입력하세요: "))

min_n, max_n = base**(N-1), base**N

answer = []

for i in range(min_n, max_n):
    # ✅ FIX: i를 문자열로 바꿔서 prefix 만들기 (A 전체를 str()로 바꾸면 안 됨)
    s = str(i)

    # ✅ FIX: 모든 prefix가 소수인지 확인 (하나라도 실패하면 탈락)
    ok = True
    for j in range(1, len(s) + 1):          # 1..N
        if not is_prime(int(s[:j])):        # s[:j] = prefix
            ok = False
            break

    # ✅ FIX: 전부 통과한 경우에만 i를 1번 추가(중복 방지)
    if ok:
        answer.append(i)

for x in answer:
    print(x)
````

````python
### 위 풀이는 전수검사라 시간복잡도 큼 O(9·10^(N-1) · N · √(10^N))
# 재귀 사용 풀이

def is_prime(x: int) -> bool:
    if x == 1: return False
    if x == 2: return True
    if x % 2 == 0: return False
    r = int(x**0.5)
    for d in range(3, r+1, 2):
        if x % d == 0:
            return False
    return True

N = int(input())

def dfs(num: int, length: int):
    if length == N:
        print(num)
        return
    for d in range(10):
        nxt = num * 10 + d
        if is_prime(nxt):
            dfs(nxt, length + 1)

# 시작은 1자리 소수들
for start in [2, 3, 5, 7]:
    dfs(start, 1)
````

# 친구관계
### 입력1: 노드 수, 에지 수
### 입력2: (에지 수 만큼) a, b => a,b가 친구관계
### depth 5이면 1 출력 + 종료

````python
import sys
sys.setrecursionlimit(10**6)

n,e = map(int,input().split())

graph = [[] for _ in range(n)]
visited = [0]*n

for _ in range(e):
    a,b = map(int, input().split())
    graph[a].append(b); graph[b].append(a)

def dfs(x, depth):
    if depth == 5:
        return 1
    visited[x] = 1
    for next in graph[x]:
        if visited[next] == 0:
            if dfs(next,depth+1)==1:
                return 1
    ## 이 반복문을 빠져나왔다는건, graph[x]에서 성사가 안됐다는거임.
    visited[x] = 0
    return 0

for i in range(n):
    if dfs(i,1) == 1:
        print(1);sys.exit()
print(0)
````
