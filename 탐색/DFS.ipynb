````python
N,E = map(int,input().split())
visited = [0]*(N+1)
A = [[] for _ in range(N+1)] ## [[],[],[],[],...]
for i in range(E):
    a,b = map(int,input().split())
    ## 양방향 엣지라 양쪽에 다 
    A[a].append(b)
    A[b].append(a)

def DFS(v):
    visited[v] = 1
    for i in A[v]:
        if not visited[i]:
            DFS(i)
count = 0
for i in range(1,N+1): ## 1~n까지 수행
    if visited[i] != 1:
        count+=1
        DFS(i)
print(count)
````

### 신기한 소수 찾기
숫자 N이 주어졌을 때, N자리 소수 중 신기한 소수 모두 찾기 (1<= N <= 8)

````python
# 소수 판별 함수
def is_prime(i: int) -> bool:
    if i < 2:
        return False
    if i == 2:
        return True
    if i % 2 == 0:
        return False
    r = int(i**0.5)
    for d in range(3, r + 1, 2):
        if i % d == 0:
            return False
    return True

base = 10
N = int(input("자릿수 입력하세요: "))

min_n, max_n = base**(N-1), base**N

answer = []

for i in range(min_n, max_n):
    # ✅ FIX: i를 문자열로 바꿔서 prefix 만들기 (A 전체를 str()로 바꾸면 안 됨)
    s = str(i)

    # ✅ FIX: 모든 prefix가 소수인지 확인 (하나라도 실패하면 탈락)
    ok = True
    for j in range(1, len(s) + 1):          # 1..N
        if not is_prime(int(s[:j])):        # s[:j] = prefix
            ok = False
            break

    # ✅ FIX: 전부 통과한 경우에만 i를 1번 추가(중복 방지)
    if ok:
        answer.append(i)

for x in answer:
    print(x)
````

````python
### 위 풀이는 전수검사라 시간복잡도 큼 O(9·10^(N-1) · N · √(10^N))
# 재귀 사용 풀이

def is_prime(x: int) -> bool:
    if x == 1: return False
    if x == 2: return True
    if x % 2 == 0: return False
    r = int(x**0.5)
    for d in range(3, r+1, 2):
        if x % d == 0:
            return False
    return True

N = int(input())

def dfs(num: int, length: int):
    if length == N:
        print(num)
        return
    for d in range(10):
        nxt = num * 10 + d
        if is_prime(nxt):
            dfs(nxt, length + 1)

# 시작은 1자리 소수들
for start in [2, 3, 5, 7]:
    dfs(start, 1)
````

# 친구관계
### 입력1: 노드 수, 에지 수
### 입력2: (에지 수 만큼) a, b => a,b가 친구관계
### depth 5이면 1 출력 + 종료

````python
import sys
sys.setrecursionlimit(10**6)

n,e = map(int,input().split())

graph = [[] for _ in range(n)]
visited = [0]*n

for _ in range(e):
    a,b = map(int, input().split())
    graph[a].append(b); graph[b].append(a)

def dfs(x, depth):
    if depth == 5:
        return 1
    visited[x] = 1
    for next in graph[x]:
        if visited[next] == 0:
            if dfs(next,depth+1)==1:
                return 1
    ## 이 반복문을 빠져나왔다는건, graph[x]에서 성사가 안됐다는거임.
    visited[x] = 0
    return 0

for i in range(n):
    if dfs(i,1) == 1:
        print(1);sys.exit()
print(0)
````

````python
import threading
import time
import random

BUF = []
MAX_SIZE = 5

produced = 0
consumed = 0
index_errors = 0
overflow_events = 0

stop = False

def producer(pid, loops=20000):
    global produced, overflow_events
    for k in range(loops):
        item = (pid, k)

        # ❌ 레이스 포인트: "길이 확인"과 "append"가 한 덩어리가 아님
        if len(BUF) < MAX_SIZE:
            time.sleep(0)  # 일부러 스위칭 유도
            BUF.append(item)

            # ❌ 레이스 포인트: produced += 1 도 원자적이지 않음(값 누락 가능)
            produced += 1
        else:
            time.sleep(0)

        # (옵션) 더 랜덤하게 흔들기
        if k % 1000 == 0:
            time.sleep(0)

def consumer(cid):
    global consumed, index_errors
    while True:
        if stop and not BUF:
            return

        # ❌ 레이스 포인트: "비었는지 확인"과 "pop"이 한 덩어리가 아님
        if len(BUF) > 0:
            time.sleep(0)  # 일부러 스위칭 유도
            try:
                BUF.pop(0)
            except IndexError:
                # 다른 소비자가 먼저 pop해서 비어버리면 터짐
                index_errors += 1
                continue

            consumed += 1
        else:
            time.sleep(0)

def main():
    global stop, overflow_events

    producers = [threading.Thread(target=producer, args=(p,)) for p in range(3)]
    consumers = [threading.Thread(target=consumer, args=(c,)) for c in range(5)]

    for t in consumers: t.start()
    for t in producers: t.start()

    for t in producers: t.join()
    stop = True
    for t in consumers: t.join()

    # 버퍼가 MAX_SIZE를 넘는 일도 실제로 생길 수 있음(체크-후-행동 레이스)
    if len(BUF) > MAX_SIZE:
        overflow_events += 1

    print("produced =", produced)
    print("consumed =", consumed)
    print("left in BUF =", len(BUF))
    print("IndexError count =", index_errors)
    print("BUF overflow? (len>MAX) =", len(BUF) > MAX_SIZE)

if __name__ == "__main__":
    main()
````

## Back Tracking
### O(N^d)
Q. N,M 주어졌을 때, 1~N 자연수 중 중복없이 M개 고른 수열 

````python
N,M = map(int,input().split())

S = [0]*M ## 수열 저장할 리스트
visited = [0]*N ## 숫자 사용 여부 저장할 리스트

def backtrack(length):
    ## 정답일 경우 수행할 내용
    if length == M:
        print(' '.join(str(x+1) for x in S)) ## 1~N인데 S는 0~N-1이니까 +1 해줘야함. 
        return
    ## 아직 정답이 아닐 경우 (length < M)
    for i in range(N):
        if not visited[i]:
            ## i번째 방문처리
            visited[i] = 1
            S[length] = i
            ## 길이 + 1
            backtrack(length+1)
            ## 위를 지나왔다면, i를 방문하는 경우들은 다 처리됨.
            ## 수 반납
            visited[i] = 0
backtrack(0)
````

````python
# N-Queen
## N*N 체스판에 퀸 N개를 서로 공격할 수 없게 놓기
# 00 01 02 03 
# 10 11 12 13
# 20 21 22 23
# 30 31 32 33

## row - col이 같으면 같은 ↘
## row + col이 같으면 같은 ↗

import sys
sys.setrecursionlimit(10**6)

N = int(input())
cnt = 0

cols = [0]*N
diag1 = [0]*(2*N-1) # row+col :: 0 ~ 2N-1
diag2 = [0]*(2*N-1) # row -col + N - 1

def backtrack(row):
    global cnt
    if row == N: ## row가 N이라는 건 0,1,2,3,....N-1까지 배치 다 했다는거임.
        cnt+= 1
        return
    for col in range(N):
        if not cols[col] and not diag1[row+col] and not diag2[row-col+N-1]:
            ## 현재 row와 col에 놓기 -> 겹치면 안되는 세 배열에 표시
            cols[col] = diag1[row + col] = diag2[row-col+N-1] = 1
            backtrack(row+1) ## row + 1에서도 수행. --- 정답까지 간다면, 재귀 중 row==N 만나서 리턴될거임.
            ## 위를 벗어났다면, 현재 col에서 정답이 되는게 없다는 말임. 표시한거 다 지우기
            cols[col] = diag1[row + col] = diag2[row-col+N-1] = 0
backtrack(0)
print(cnt)
````
