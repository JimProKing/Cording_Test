N,M 미로가 주어질 때, (1,1)에서 출발해 N,M까지 지나야 하는 칸 수의 최솟값

- DFS: 한 방향으로 끝까지 들어감

- BFS: 시작점에서 거리 1인 칸들 전부, 그다음 거리 2인 칸들 전부 이런 식으로 감

```python
from collections import deque

dx = [0,1,0,-1]
dy = [1,0,-1,0]

N, M = map(int, input().split())
A = [[0]*M for _ in range(N)]
visited = [[False]*M for _ in range(N)]

for i in range(N):
    numbers = list(input().strip())
    for j in range(M):
        A[i][j] = int(numbers[j])

def BFS(i, j):
    queue = deque()
    queue.append((i, j))
    visited[i][j] = True

    while queue:
        now = queue.popleft()
        for k in range(4):
            x = now[0] + dx[k]
            y = now[1] + dy[k]

            if 0 <= x < N and 0 <= y < M:
                if A[x][y] != 0 and not visited[x][y]:
                    visited[x][y] = True
                    A[x][y] = A[now[0]][now[1]] + 1
                    queue.append((x, y))

BFS(0, 0)
print(A[N-1][M-1])
```
