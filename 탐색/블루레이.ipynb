Mission: 블루레이 크기 중 최솟값 구하는 프로그램
- 블루레이: 녹화할 수 있는 길이
- 시작 인덱스: 최대 길이 레슨
- 종료 인덱스: 모든 레슨 길이의 합

* 이진탐색 수행하면서, 3번째 탐색에 시작인덱스>종료인덱스 조건 만족할 때 이진탐색 종료 시 시작인덱스가 답

```python
N,M = map(int,input().split())
A = list(map(int,input().split())) ## 분단위 블루레이
start = 0
end = 0

for i in A:
    if start < i:
        start = i
    end += i

while start <= end:
    middle = int((start+end)/2)
    sum = 0; count = 0
    for i in range(N):
        if sum+A[i] > middle:
            count+=1
            sum = 0
        sum += A[i]
    if sum != 0:
        count+=1
    if count > M:
        start = middle + 1
    else:
        end = middle - 1

print(start)
```
